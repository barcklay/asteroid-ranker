```python
import pandas as pd
import json

# JPL CSV (на самом деле JSON)
with open('/Users/elzira/my-proj/hardware/asteroid-ranker/data/neas.csv') as f:
    raw = json.load(f)

print("Fields:", raw['fields'])
print("Total objects:", len(raw['data']))

df = pd.DataFrame(raw['data'], columns=raw['fields'])
df.head()
```

```python
# Преобразуем числовые колонки (они приходят строками из JSON)
numeric_cols = ['H', 'diameter', 'albedo', 'e', 'a', 'q', 'i', 'om', 'w', 'per', 'moid']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print(f"Total NEAs: {len(df)}")
print(f"With diameter: {df['diameter'].notna().sum()}")
print(f"With moid: {df['moid'].notna().sum()}")
print(f"PHAs (potentially hazardous): {(df['pha'] == 'Y').sum()}")
df.describe()
```
